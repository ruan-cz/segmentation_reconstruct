```python
import numpy as np
from plyfile import PlyData, PlyElement

models_name = [
    # todo
    # "00025183_021c990acfc648618a49bd47_trimesh_000",
    # "00025611_37f3a717a9b842cfbe5f6005_trimesh_002",
    # "00025633_2aaf78a0f41a47b1bae8c447_trimesh_003",
    # "00025710_6fe81cf35e2740b3bbde9aa6_trimesh_020",
    # "00025728_6fe81cf35e2740b3bbde9aa6_trimesh_038",
    # "00025749_c01a880ceccb4e05908f8255_trimesh_000",
    # "00026055_11394f3abb83419b88f6eed9_trimesh_000",
    # "00026133_8010d7708c06499e9e99ce1c_trimesh_008",

    # "00140400_e4a5e7862a9beda824b2e00c_trimesh_003",
    # "00140553_e2c0841b6c86e3bfdcc8c477_trimesh_000",
    # '00140710_33073a6cb55fadd20ec0e3fb_trimesh_001',
    # "00140750_0da4d13f288d4cd70deecf20_trimesh_001",
    # "00140756_eaf6606df2a105f356a99677_trimesh_000",
    # "00140919_3023c3cfd7d6b1a80700e17e_trimesh_000",
    # "00140923_0e818d7ff8780682567c824e_trimesh_000",
    # "00140936_9a2b76f7e489ea90f5e41ef8_trimesh_007",
    # "00140967_631426bb5199fa39fc8eeb1c_trimesh_008",
    # "00140982_f2e481e34cff3a450799acac_trimesh_000",
    # "00141097_57222034e4b08229a097b208_trimesh_010",
]

import trimesh
import numpy as np
def align_mesh_by_bbox(src_path, tgt_path, out_path):
    """
    将 src_path (2.ply) 对齐到 tgt_path (1.obj / 1.off)
    """
    src = trimesh.load(src_path, process=False)
    tgt = trimesh.load(tgt_path, process=False)

    V_src = src.vertices
    V_tgt = tgt.vertices

    # bounding box
    min_src, max_src = V_src.min(axis=0), V_src.max(axis=0)
    min_tgt, max_tgt = V_tgt.min(axis=0), V_tgt.max(axis=0)

    center_src = (min_src + max_src) / 2
    center_tgt = (min_tgt + max_tgt) / 2

    size_src = max_src - min_src
    size_tgt = max_tgt - min_tgt

    # scale（三种选法）
    scale_xyz = size_tgt / size_src
    scale = scale_xyz.mean()      # 推荐：整体统一缩放
    # scale = scale_xyz            # 若你想逐轴缩放

    # apply transform
    V_new = (V_src - center_src) * scale + center_tgt
    src.vertices = V_new

    src.export(out_path)
    print(f"Saved aligned mesh to {out_path}")


for model_name in models_name:

    raw_off = f"example_data/selected/{model_name}.obj"
    # raw_off = f"example_data/abc_all/meshes/{model_name}.off"
    partfield_ply = f"example_results/merge/selected/example_result/partfield/{model_name}.ply"
    partfield_ply_new = f"example_results/merge/selected/example_result/partfield_trans/{model_name}.ply"
    segmat_ply = f"example_results/merge/selected/example_result/segmat/{model_name}.ply"
    segmat_ply_new = f"example_results/merge/selected/example_result/segmat_trans/{model_name}.ply"
    samesh_ply = f"example_results/merge/selected/example_result/samesh/{model_name}.ply"
    samesh_ply_new = f"example_results/merge/selected/example_result/samesh_trans/{model_name}.ply"

    align_mesh_by_bbox(
        partfield_ply, raw_off, partfield_ply_new
    )
    align_mesh_by_bbox(
        segmat_ply, raw_off, segmat_ply_new
    )
    align_mesh_by_bbox(
        samesh_ply, raw_off, samesh_ply_new
    )


```
