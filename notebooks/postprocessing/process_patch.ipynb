```python
obj_name = '2'
# obj_name = '1_split'
# tmp_path = f'example_results/merge/selected/example_result/00140223_5374ce893909eddbc7536b71_trimesh_000_our/{obj_name}.ply'
ply_our_path = f'example_results/merge/selected/example_result/our/{obj_name}.ply'
ply_p2s_path = f'example_results/merge/selected/example_result/part2surf/{obj_name}.ply'
ply_samesh_path = f'example_results/merge/selected/example_result/samesh/{obj_name}.ply'
ply_sdf_path = f'example_results/merge/selected/example_result/sdf/{obj_name}.ply'

file = open('utils/colors_500.txt', 'r')
lines = file.readlines()
colors = np.zeros((len(lines),3))
for i in range(len(lines)):
    colors[i] = np.array(lines[i].split())

def save_patch(ply_path, method):
    # read v, f, f_colors from ply_path
    mesh = trimesh.load(ply_path, process=False)
    v = mesh.vertices
    f = mesh.faces
    f_colors = mesh.visual.face_colors[:,:3]

    # cluster f by color
    unique_colors = np.unique(f_colors, axis=0)
    patch2f = {}
    for i, c in enumerate(unique_colors):
        patch2f[i] = np.where((f_colors == c).all(axis=1))[0]

    output_folder = f'example_results/merge/selected/example_result/{obj_name}_{method}/'
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    # save f in each patch to obj
    for p_id in range(len(patch2f)):
        f_patch = f[patch2f[p_id]]
        f_colors_patch = np.ones((f_patch.shape[0], 3)) * colors[p_id % 100]
        write_ply_file(v, f_patch, f_colors_patch, 
            os.path.join(output_folder, f'{p_id}.ply')
        )
# save_patch(tmp_path, 'our')
save_patch(ply_our_path, 'our')
# save_patch(ply_samesh_path, 'samesh')

```
