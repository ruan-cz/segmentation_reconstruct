```python
import numpy as np

def dedup_vertices(v0, f0):
    """
    Deduplicate identical vertices and remap faces.

    Parameters
    ----------
    v0 : (N,3) float array
        Original vertices
    f0 : (M,K) int array, 0-based
        Original faces

    Returns
    -------
    v1 : (N',3) float array
        Deduplicated vertices
    f1 : (M,K) int array, 0-based
        Remapped faces
    """
    v0 = np.asarray(v0)
    f0 = np.asarray(f0)

    # unique vertices
    v1, inverse = np.unique(v0, axis=0, return_inverse=True)
    # inverse[i] = new index of old vertex i

    # remap faces
    f1 = inverse[f0]

    return v1, f1



```
