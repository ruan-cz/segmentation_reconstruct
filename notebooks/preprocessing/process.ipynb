```python
import numpy as np
import igl
import meshplot as mp
import openmesh as om
import networkx as nx
import trimesh
from collections import defaultdict
import time
import open3d as o3d



obj_name = '00140750_0da4d13f288d4cd70deecf20_trimesh_001_1_1_1_1'
obj_path = f'example_results\merge\selected/{obj_name}.obj'
output_path = 'example_results\merge\selected'
ply_path = f'example_results\merge\selected/{obj_name}.ply'

file = open('utils/colors_500.txt', 'r')
lines = file.readlines()
colors = np.zeros((len(lines),3))
for i in range(len(lines)):
    colors[i] = np.array(lines[i].split())

def set_ply(obj_path, ply_path):
    v, f = igl.read_triangle_mesh(obj_path)
    f_colors = np.ones((f.shape[0], 3), dtype=np.float32)
    for i in range(f.shape[0]):
        f_colors[i] = colors[61]
    mesh = trimesh.Trimesh(v, f)
    mesh.visual.face_colors = (f_colors * 255).astype(np.uint8)
    mesh.export(ply_path, file_type='ply')
# set_ply(obj_path, ply_path)


def split_mesh(obj_path):
    mesh = trimesh.load(obj_path, force='mesh')
    components = mesh.split(only_watertight=False)
    return components
def write_components(components, output_path):
    for i, component in enumerate(components):
        component.export(f"{output_path}/{obj_name}_part_{i}.obj")
        print(f"Component {i} saved to {output_path}/component_{i}.obj")


obj0 = '00140780_d56b3836d8a40e3035d42380_trimesh_000_part_1_proxy'
obj1 = '00140780_d56b3836d8a40e3035d42380_trimesh_000_0_proxy'
obj2 = '00140780_d56b3836d8a40e3035d42380_trimesh_000_part_0_part_1_proxy'

# merge ply file into one file
def merge_ply_files(output_path, obj0, obj1, obj2):
    mesh0 = trimesh.load(f"{output_path}/{obj0}.ply")
    mesh1 = trimesh.load(f"{output_path}/{obj1}.ply")
    mesh2 = trimesh.load(f"{output_path}/{obj2}.ply")

    merged_mesh = trimesh.util.concatenate([mesh0, mesh1, mesh2])
    merged_mesh.export(f"{output_path}/00140780_d56b3836d8a40e3035d42380_trimesh_000.ply", file_type='ply')
    print(f"Merged mesh saved to {output_path}/merged_mesh.ply")
merge_ply_files(output_path, obj0, obj1, obj2)

```
