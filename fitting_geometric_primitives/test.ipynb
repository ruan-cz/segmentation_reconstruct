```python
def read_pcd_from_txt(file):
    file = open(file, 'r')
    lines = file.readlines()
    points = []
    for line in lines:
        line = line.strip().split(',')
        if len(line) == 3:
            points.append([float(line[0]), float(line[1]), float(line[2])])
    return np.asarray(points)
    
    
def show_pcd(points):
    print('Pointcloud shape: ', points.shape)
    mp.plot(points, shading={"point_size": 0.2})
    

def fit(file):
    pcd = read_pcd_from_txt(file)
    show_pcd(pcd)
    fitting = PrimitiveFitting(v=pcd, f=None)
    v_id = range(pcd.shape[0])
    print('Plane: ', fitting.fit_planar(v_id, is_v_id=True))
    print('Sphere: ', fitting.fit_sphere(v_id, is_v_id=True))
    print('Cylinder: ', fitting.fit_cylinder(v_id, is_v_id=True))
    print('Cone: ', fitting.fit_cone(v_id, is_v_id=True))
    # print('Torus: ', fitting.fit_torus(v_id, is_v_id=True))
    

for i in range(1, 2):
    print('===Fitting pointcloud {}==='.format(i))
    file_path =  'test/pointCloud/pointCloud{}.txt'.format(i)
    fit(file_path)
```
