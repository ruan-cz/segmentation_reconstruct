```python
def test_generate_sphere():
    # generate sphere pointcloud
    n_sample = 10
    theta = np.linspace(0, np.pi, n_sample)
    phi = np.linspace(0, np.pi, n_sample)
    
    x = np.zeros(n_sample**2)
    y = np.zeros(n_sample**2)
    z = np.zeros(n_sample**2)
    for i in range(n_sample):
        for j in range(n_sample):
            x[i*n_sample+j] = np.sin(theta[i]) * np.cos(phi[j])
            y[i*n_sample+j] = np.sin(theta[i]) * np.sin(phi[j])
            z[i*n_sample+j] = np.cos(theta[i])
    x += np.random.normal(0, 0.01, n_sample**2)
    y += np.random.normal(0, 0.01, n_sample**2)
    z += np.random.normal(0, 0.01, n_sample**2)
    pointcloud = np.array([x, y, z]).T
    print(pointcloud.shape)
    
    # fig = plt.figure()
    # ax = fig.add_subplot(111, projection='3d')
    # ax.scatter(x, y, z)
    # plt.show()
    
    sphere = Sphere(pointcloud, None, None, None, None)
    fit_rate, fit_params = sphere.fit(np.arange(n_sample**2), True)
    print(fit_rate, fit_params)
    
    # Taubin
    # quadric_surface = QuadricSurface()
    # params = quadric_surface.fit(pointcloud)
    # print(params)
    
    
    # lib
    sphere = pyrsc.Sphere()
    center, radius, _ = sphere.fit(pointcloud, thresh=0.4)
    print(center, radius)


test_generate_sphere()
    
```

```python
def test_generate_cone():
    apex = np.array([0, 0, 0])
    theta_axis = np.random.rand() * np.pi
    phi_axis = np.random.rand() * np.pi
    axis = np.array([
        np.sin(theta_axis) * np.cos(phi_axis),
        np.sin(theta_axis) * np.sin(phi_axis),
        np.cos(theta_axis)
    ])
    theta = np.random.rand() * 0.5 * np.pi
    h_max = 1
    n_height = 10
    n_circle = 10
    
    # craete uv basis
    arb = np.array([1., 0., 0.])
    if np.abs(np.dot(abs, axis)) > 0.9:
        arb = np.array([0., 1., 0.])
    u = arb - np.dot(arb, axis) * axis
    u /= np.linalg.norm(u)
    v = np.cross(u, axis)
    
    p_apex = apex.reshape(1, 3)
    h_list = np.linspace(h_max / n_height, h_max, n_height)
    h_centers = apex + h_list[:, np.newaxis] * axis
    radii = h_list * np.tan(theta)
    phi_list = np.linspace(0, 2 * np.pi, n_circle, endpoint=False)
    
    radial_vec = np.cos(phi_list)[np.newaxis, :, np.newaxis] * u + np.sin(phi_list)[np.newaxis, :, np.newaxis] * v
    radial_vec *= radii[:, np.newaxis, np.newaxis]
    
    surface_points = h_centers[:, np.newaxis, :] + radial_vec
    surface_points = surface_points.reshape(-1, 3)
    
    points = np.vstack([p_apex, surface_points])
    
    
    fig = plt.figure()
    ax = fig.add_subplot(111, projection='3d')
    ax.scatter(points[:, 0], points[:, 1], points[:, 2])
    plt.show()
    
    print(apex, axis, theta)
    
    # fit 
    cone = Cone(points, None, None)
    

    
    
        
    

    
```
