```python
import numpy
from collections import defaultdict
import igl

from utils import *


def convert_proxy_file(
    input_file,
    output_file,
    face_num
):
    face_to_proxy = [-1] * face_num

    with open(input_file, "r") as f:
        for proxy_id, line in enumerate(f):
            if not line.strip():
                continue
            face_ids = map(int, line.strip().split())
            for face_id in face_ids:
                if face_id < 0 or face_id >= face_num:
                    raise ValueError(f"Invalid face id {face_id}")
                face_to_proxy[face_id] = proxy_id

    with open(output_file, "w") as f:
        for proxy in face_to_proxy:
            f.write(f"{proxy}\n")

def invert_face_to_proxy(
    input_file,
    output_file
):
    proxy_to_faces = defaultdict(list)

    with open(input_file, "r") as f:
        for face_id, line in enumerate(f):
            proxy_id = int(line.strip())
            if proxy_id == -1:
                continue
            proxy_to_faces[proxy_id].append(face_id)

    if not proxy_to_faces:
        raise RuntimeError("No valid proxy found")

    max_proxy_id = max(proxy_to_faces.keys())

    with open(output_file, "w") as f:
        for proxy_id in range(max_proxy_id + 1):
            faces = proxy_to_faces.get(proxy_id, [])
            if faces:
                f.write(" ".join(map(str, faces)))
            f.write("\n")

def convert_pid_ply(
    pid_file,
    obj_file,
    ply_file
):
    v, f = igl.read_triangle_mesh(obj_file)
    face_num = f.shape[0]
    face_to_proxy = [-1] * face_num
    # print(face_num)
    with open(pid_file, "r") as f_pid:
        for face_id, line in enumerate(f_pid):
            proxy_id = int(line.strip())
            face_to_proxy[face_id] = proxy_id

    face_colors = np.ones((face_num, 3))
    for face_id, proxy_id in enumerate(face_to_proxy):
        face_colors[face_id] = colors[proxy_id]
    write_ply_file(v, f, face_colors, ply_file)

import os
import shutil
def copy_pid_to_seg(name, method):
    folder = "D:/Desktop/MeshSegment/"
    src = os.path.join(folder, name)
    base = name[:-len(".obj.pid")]
    dst_name = base + f"_{method}.seg"
    dst = os.path.join(folder, dst_name)
    shutil.copyfile(src, dst)
    print(f"{name} -> {dst_name}")

def copy_seg_to_pid(name, method):
    folder = "D:/Desktop/MeshSegment/"
    base = name[:-len(f".obj.pid")]
    src = os.path.join(folder, name[:-len(f".obj.pid")]+f"_{method}.seg")
    dst_name = base + ".obj.pid"
    dst = os.path.join(folder, dst_name)
    print(f"{src} -> {dst}")
    shutil.copyfile(src, dst)
```
