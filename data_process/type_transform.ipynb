```python
# transform obj mesh in folder example_data/selected to off
import os
import trimesh

input_folder = '../example_data/selected/'
output_folder = '../example_data/selected_off/'
if not os.path.exists(output_folder):
    os.makedirs(output_folder) 
for filename in os.listdir(input_folder):
    if filename.endswith('.obj'):
        mesh = trimesh.load(os.path.join(input_folder, filename))
        mesh.export(os.path.join(output_folder, filename.replace('.obj', '.off')))
        print(f'Converted {filename} to OFF format.')
```
