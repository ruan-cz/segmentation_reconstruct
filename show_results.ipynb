```python
import numpy as np
import igl
import meshplot as mp
import openmesh as om
import networkx as nx
import trimesh
from collections import defaultdict
import os

%load_ext autoreload
%autoreload 2
# import sys
# sys.path.append('..')
from utils import *
from primitive_fitting import *
from hole_filling import *
from cut_calculate.intersection import *


file = open('global/colors_500.txt', 'r')
lines = file.readlines()
colors = np.zeros((len(lines),3))
for i in range(len(lines)):
    colors[i] = np.array(lines[i].split())
```

```python
obj_name = '2'
obj_file = f'example_data/example_CAD/2.obj'
path = 'example_results/vsa'
cluster_path = os.path.join(path, f'{obj_name}_proxy_cluster.txt')
region_path = os.path.join(path, f'{obj_name}_region_cluster.txt')
merged_cluster_path = os.path.join(path, f'{obj_name}_merged_proxy.txt')

v, f = igl.read_triangle_mesh(obj_file)

cluster2f, region2cluster = read_regioned_cluster(cluster_path, region_path)


print('region cluster')
f_color = np.zeros((f.shape[0], 3))
for i in range(len(region2cluster)):
    clusters = region2cluster[i]
    for cluster in clusters:
        f_color[cluster2f[cluster]] = colors[i % 100]
mp.plot(v, f, c=f_color, shading={'wireframe': True}, return_plot=True)


print('raw cluster')
f_color = np.zeros((f.shape[0], 3))
for i in range(len(cluster2f)):
    f_color[cluster2f[i]] = colors[i % 100]
mp.plot(v, f, c=f_color, shading={'wireframe': True}, return_plot=True)


print('merged cluster')
merged_cluster = []
mc_f = open(merged_cluster_path, 'r')
for line in mc_f.readlines():
    merged_cluster.append(np.array(line.split(), dtype=int))
f_color = np.zeros((f.shape[0], 3))
for i in range(len(merged_cluster)):
    clusters = merged_cluster[i]
    for cluster in clusters:
        f_color[cluster2f[cluster]] = colors[i % 100]
mp.plot(v, f, c=f_color, shading={'wireframe': True}, return_plot=True)


```
